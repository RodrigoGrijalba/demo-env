```python
import pandas as pd
import numpy as np
from zipfile import ZipFile
import os
```

```python
INPUT_DIR = "../../../raw/"
OUTPUT_DIR = "../../../clean"
DEMOCRACY_COLUMNS: list[str] = [
        "country_name", "wbcode", "wbcode2", "year", "dem", "PopulationtotalSPPOPTOTL",
        "gdppercapitaconstant2000us", "_ID"
]
```

```python
def clean_ddcg():
        raw_ddcg_path = os.path.join(INPUT_DIR, "DDCGdata_final.dta")
        table = pd.read_stata(raw_ddcg_path)[DEMOCRACY_COLUMNS]
        table = (
                table
                .rename(columns={"gdppercapitaconstant2000us": "gdp_pc"})
        )
        clean_ddcg_path = os.path.join(OUTPUT_DIR, "DDCGdata.csv")
        table.to_csv(clean_ddcg_path, index=False)

def main():
        clean_ddcg()
```

```python
if __name__ == "__main__":
        main()
```

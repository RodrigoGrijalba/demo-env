```python
import pandas as pd
import numpy as np
from zipfile import ZipFile
import os
```

```python
INPUT_DIR = "../../../raw/"
OUTPUT_DIR = "../../../clean"
COUNTRY_RENAME_MAPPING = {
        "NRU": "NAU", "ROU": "ROM", "SCG": "SER", "SGP": "SIN",
        "TWN": "TAW", "ARE": "UAE", "COD": "ZAR", "SRB": "SER",
        "MNE": "SER"
}
```

```python
def clean_wb_regions():
        raw_csv_path = os.path.join(INPUT_DIR, "wb-regions.csv")
        table = pd.read_csv(raw_csv_path)
        table["wb_region2"] = pd.factorize(table["wb_region"])[0]
        table = (
                table
                .replace(COUNTRY_RENAME_MAPPING)
                .drop(columns=["Entity", "Year", "wb_region"])
                .sort_values("Code")
                .rename(columns={"Code": "wbcode"})
        )
        clean_csv_path = os.path.join(
                OUTPUT_DIR, "wb-regions.csv"
        )
        table.to_csv(clean_csv_path, index=False)
        return

def main():
        clean_wb_regions()
```

```python
if __name__ == "__main__":
        main()
```

```python
import os
import pandas as pd
import geopandas as gpd
```

```python
INPUT_DIR = "../../../clean/"
OUTPUT_DIR = "../../../processed"
```

```python
ddcg_path = os.path.join(INPUT_DIR, "DDCGdata.csv")
ddcg = pd.read_csv(ddcg_path)
```

```python
vcf = pd.read_csv(os.path.join(INPUT_DIR, "vcf.csv"))
```

```python
vcf.country_name.unique().shape
```

```python
vcf_ctrys = set(vcf.country_name)
ddcg_ctrys = set(ddcg.country_name)
```

```python
print("Unmatched territories from vcf:")
print(list(vcf_ctrys - ddcg_ctrys), "\n")
```

```python
print("Unmatched territories from DDCG:")
print(ddcg_ctrys - vcf_ctrys, "\n")
```

```python
merged = ddcg.merge(
        vcf, "left", ["country_name", "year"]
)
```

```python
wbregions = pd.read_csv(
        os.path.join(INPUT_DIR, "wb-regions.csv")
)
```

```python
merged = (
        merged
        .merge(wbregions, "left")
        .drop_duplicates(subset=["wbcode", "year"])
)
```

```python
merged.to_stata(os.path.join(OUTPUT_DIR, "tree_cover.dta"), write_index=False)
```
